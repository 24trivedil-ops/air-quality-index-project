```python
pm25_aqi = pm25.withColumn(
    "PM25_AQI",
    when(col("Value") <= 12.0,
         col("Value") * 50 / 12.0)

    .when(col("Value") <= 35.4,
         50 + (col("Value") - 12.1) * 50 / (35.4 - 12.1))

    .when(col("Value") <= 55.4,
         100 + (col("Value") - 35.5) * 50 / (55.4 - 35.5))

    .when(col("Value") <= 150.4,
         150 + (col("Value") - 55.5) * 50 / (150.4 - 55.5))

    .when(col("Value") <= 250.4,
         200 + (col("Value") - 150.5) * 100 / (250.4 - 150.5))

    .when(col("Value") <= 350.4,
         300 + (col("Value") - 250.5) * 100 / (350.4 - 250.5))

    .otherwise(500)
)
```

```python
pm25_aqi = pm25_aqi.withColumn(
    "PM25_AQI",
    round(col("PM25_AQI"), 0)
)
```

```python
pm25_aqi = pm25_aqi.withColumn(
    "AQI_Category",

    when(col("PM25_AQI") <= 50, "Good")

    .when(col("PM25_AQI") <= 100, "Moderate")

    .when(col("PM25_AQI") <= 150, "Unhealthy for Sensitive Groups")

    .when(col("PM25_AQI") <= 200, "Unhealthy")

    .when(col("PM25_AQI") <= 300, "Very Unhealthy")

    .otherwise("Hazardous")
)
```

```python
import gradio as gr
import pandas as pd
import plotly.express as px
```

```python
country_pd = country_aqi.toPandas()
city_pd = city_aqi.toPandas()
aqi_pd = pm25_aqi.toPandas()
```

```python
print(country_pd.shape)
print(city_pd.shape)
print(aqi_pd.shape)
```

```python
countries = sorted(
    country_pd["Country Label"]
    .dropna()
    .unique()
    .tolist()
)

cities = sorted(
    city_pd["City"]
    .dropna()
    .unique()
    .tolist()
)
```

```python
import builtins

def country_analysis(country):

    data = country_pd[
        country_pd["Country Label"] == country
    ]

    if data.empty:
        return (
            "No Data",
            "No Data",
            "No Data",
            "No Data",
            pd.DataFrame()
        )

    row = data.iloc[0]

    avg_aqi = row["Average_AQI"]

    if avg_aqi <= 50:
        category = "Good"
    elif avg_aqi <= 100:
        category = "Moderate"
    elif avg_aqi <= 150:
        category = "Unhealthy for Sensitive Groups"
    elif avg_aqi <= 200:
        category = "Unhealthy"
    elif avg_aqi <= 300:
        category = "Very Unhealthy"
    else:
        category = "Hazardous"

    return (
        builtins.round(avg_aqi, 2),
        builtins.round(row["Maximum_AQI"], 2),
        builtins.round(row["Minimum_AQI"], 2),
        category,
        data
    )
```

```python
import builtins

def city_analysis(city):

    data = city_pd[
        city_pd["City"] == city
    ]

    if data.empty:
        return (
            "No Data",
            "No Data",
            "No Data",
            pd.DataFrame()
        )

    row = data.iloc[0]

    avg_aqi = row["Average_AQI"]

    if avg_aqi <= 50:
        category = "Good"
    elif avg_aqi <= 100:
        category = "Moderate"
    elif avg_aqi <= 150:
        category = "Unhealthy for Sensitive Groups"
    elif avg_aqi <= 200:
        category = "Unhealthy"
    elif avg_aqi <= 300:
        category = "Very Unhealthy"
    else:
        category = "Hazardous"

    return (
        builtins.round(avg_aqi, 2),
        builtins.round(row["Maximum_AQI"], 2),
        category,
        data
    )
```

```python
def top_country_chart():

    data = country_pd.sort_values(
        "Average_AQI",
        ascending=False
    ).head(15)

    fig = px.bar(
        data,
        x="Country Label",
        y="Average_AQI",
        title="Top 15 Countries by Average PM2.5 AQI",
        labels={
            "Country Label": "Country",
            "Average_AQI": "Average AQI"
        }
    )

    fig.update_layout(
        xaxis_tickangle=-45
    )

    return fig
```

```python
def top_city_chart():

    data = city_pd.sort_values(
        "Average_AQI",
        ascending=False
    ).head(15)

    fig = px.bar(
        data,
        x="City",
        y="Average_AQI",
        title="Top 15 Cities by Average PM2.5 AQI",
        labels={
            "City": "City",
            "Average_AQI": "Average AQI"
        }
    )

    fig.update_layout(
        xaxis_tickangle=-45
    )

    return fig
```

```python
def category_chart():

    data = aqi_pd.copy()

    data["Category"] = pd.cut(
        data["PM25_AQI"],
        bins=[-1, 50, 100, 150, 200, 300, 500],
        labels=[
            "Good",
            "Moderate",
            "Unhealthy for Sensitive Groups",
            "Unhealthy",
            "Very Unhealthy",
            "Hazardous"
        ]
    )

    category_count = (
        data["Category"]
        .value_counts()
        .reset_index()
    )

    category_count.columns = [
        "Category",
        "Count"
    ]

    fig = px.bar(
        category_count,
        x="Category",
        y="Count",
        title="Global AQI Category Distribution"
    )

    fig.update_layout(
        xaxis_tickangle=-30
    )

    return fig
```

```python
def pollutant_chart():

    data = pollutant_analysis.toPandas()

    fig = px.bar(
        data,
        x="Pollutant",
        y="Records",
        title="Global Pollutant Measurements",
        labels={
            "Pollutant": "Pollutant",
            "Records": "Number of Measurements"
        }
    )

    return fig
```

```python
with gr.Blocks(
    title="Global Air Quality Analysis"
) as demo:

    gr.Markdown(
        """
        # 🌍 Global Air Quality Analysis

        ### Apache Spark + PySpark + Gradio

        Analyze global air-quality measurements,
        PM2.5-derived AQI, countries, cities and pollutants.
        """
    )

    gr.Markdown(
        """
        **Dataset:** Global Air Quality Dataset
        **Processing:** Apache Spark
        **AQI:** Derived from PM2.5 measurements
        """
    )

    with gr.Tab("🌍 Country Analysis"):

        country_dropdown = gr.Dropdown(
            choices=countries,
            label="Select Country",
            value=countries[0]
        )

        country_button = gr.Button(
            "Analyze Country"
        )

        with gr.Row():

            avg_country = gr.Number(
                label="Average AQI"
            )

            max_country = gr.Number(
                label="Maximum AQI"
            )

            min_country = gr.Number(
                label="Minimum AQI"
            )

            category_country = gr.Textbox(
                label="AQI Category"
            )

        country_table = gr.Dataframe(
            label="Country Analysis"
        )

        country_button.click(
            country_analysis,
            inputs=country_dropdown,
            outputs=[
                avg_country,
                max_country,
                min_country,
                category_country,
                country_table
            ]
        )

    with gr.Tab("🏙️ City Analysis"):

        city_dropdown = gr.Dropdown(
            choices=cities,
            label="Select City",
            value=cities[0]
        )

        city_button = gr.Button(
            "Analyze City"
        )

        with gr.Row():

            avg_city = gr.Number(
                label="Average AQI"
            )

            max_city = gr.Number(
                label="Maximum AQI"
            )

            category_city = gr.Textbox(
                label="AQI Category"
            )

        city_table = gr.Dataframe(
            label="City Analysis"
        )

        city_button.click(
            city_analysis,
            inputs=city_dropdown,
            outputs=[
                avg_city,
                max_city,
                category_city,
                city_table
            ]
        )

    with gr.Tab("📊 Visualizations"):

        gr.Markdown(
            "## Top Countries"
        )

        country_chart = gr.Plot(
            value=top_country_chart()
        )

        gr.Markdown(
            "## Top Cities"
        )

        city_chart = gr.Plot(
            value=top_city_chart()
        )

        gr.Markdown(
            "## AQI Categories"
        )

        category_plot = gr.Plot(
            value=category_chart()
        )

        gr.Markdown(
            "## Pollutants"
        )

        pollutant_plot = gr.Plot(
            value=pollutant_chart()
        )

    with gr.Tab("📋 AQI Records"):

        gr.Dataframe(
            value=aqi_pd.head(100),
            label="PM2.5 AQI Records",
            interactive=False
        )

    gr.Markdown(
        """
        ---
        ### About the project

        This dashboard uses Apache Spark to process global
        air-quality observations.

        **AQI is derived from PM2.5 concentration.**
        It is an analytical calculation and not a direct AQI
        value supplied by the original dataset.
        """
    )
```
